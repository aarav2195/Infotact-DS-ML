# Project Objective

The objective of this project is to develop a Reinforcement Learning based dynamic pricing system for the travel and hospitality industry. The pricing agent will learn an optimal pricing strategy that maximizes hotel revenue while considering customer demand and limited room inventory.

# Business Problem

Hotels have a limited number of rooms available for booking. Customer demand changes as the departure date approaches and also depends on the room price. Higher prices generate more revenue per booking but may reduce customer demand, while lower prices increase demand but reduce profit. The goal is to learn an optimal pricing strategy that balances these trade-offs.

# Markov Decision Process

State

- Remaining hotel rooms
- Days remaining until departure

Action

- Select a room price from predefined pricing levels

Reward

- Revenue generated from customer bookings

```python
TOTAL_ROOMS = 50
BOOKING_DAYS = 30

PRICE_LEVELS = [80,100,120,140,160]
```
