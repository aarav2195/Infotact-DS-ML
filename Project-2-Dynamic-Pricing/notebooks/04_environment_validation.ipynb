```python
import os
import sys

project_root = os.path.abspath("..")

if project_root not in sys.path:
    sys.path.append(project_root)

from src.environment import HotelPricingEnv
```

```python
env = HotelPricingEnv()
state,_ = env.reset()
print(state)
```

```python
import numpy as np

for episode in range(3):
    print("="*50)
    print(f"Simulation {episode+1}")
    state,_ = env.reset()
    total_reward = 0
    done = False
    while not done:
        action = np.random.randint(0,5)
        state,reward,done,_,info = env.step(action)
        total_reward += reward
    print("Revenue: ",total_reward)
    print("Rooms Remaining: ",state[0])
    print("Days Remaining: ",state[1])
```

```python
revenues = []

for _ in range(100):
    state,_ = env.reset()
    total_reward = 0
    done = False
    while not done:
        action = np.random.randint(0,5)
        state,reward,done,_,info = env.step(action)
        total_reward += reward
    revenues.append(total_reward)
print("Average Revenue: ",np.mean(revenues))
print("Maximum Revenue: ",np.max(revenues))
print("Minimum Revenue: ",np.min(revenues))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.hist(revenues,bins=15)

plt.title("Revenue Distribution")

plt.xlabel("Revenue")

plt.ylabel("Frequency")

plt.show()
```
