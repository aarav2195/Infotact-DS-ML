```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
baseline = pd.read_csv(r"..\data\processed\baseline_results.csv")
comparison = pd.read_csv(r"..\data\processed\qlearning_results.csv")
comparison
```

```python
print("="*60)
print("Business Performance Summary")
print("="*60)

best_strategy = comparison.loc[comparison["Average Revenue"].idxmax(),"Strategy"]

print(f"Best Strategy           : {best_strategy}")
print(f"Highest Revenue         : {comparison["Average Revenue"].max():.2f}")
print(f"Highest Occupancy (%)   : {comparison["Average Occupancy (%)"].max():.2f}")
print(f"Highest Avg Price       : {comparison["Average Price"].max():.2f}")
print(f"Highest Rooms Sold      : {comparison["Average Rooms Sold"].max():.2f}")
```

```python
plt.figure(figsize=(8,5))

plt.bar(comparison["Strategy"],comparison["Average Revenue"])

plt.title("Average Revenue Comparison")

plt.ylabel("Revenue")

plt.grid(axis="y")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.bar(comparison["Strategy"],comparison["Average Occupancy (%)"])

plt.title("Average Occupancy Comparison")

plt.ylabel("Occupancy (%)")

plt.grid(axis="y")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.bar(comparison["Strategy"],comparison["Average Rooms Sold"])

plt.title("Average Rooms Sold Comparison")

plt.ylabel("Rooms Sold")

plt.grid(axis="y")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.bar(comparison["Strategy"],comparison["Average Price"])

plt.title("Average Room Price Comparison")

plt.ylabel("Price")

plt.grid(axis="y")

plt.show()
```
