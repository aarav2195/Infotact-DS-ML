```python
agent = DQNAgent(state_size=2,action_size=len(PRICE_LEVELS))

print(agent)
```

```python
print("=" * 60)
print("Agent Information")
print("=" * 60)
print("State Size :", agent.state_size)
print("Action Size:", agent.action_size)
print("Gamma      :", agent.gamma)
print("Learning Rate :", agent.learning_rate)
print("Epsilon :", agent.epsilon)
print("Replay Buffer Size :", len(agent.memory))
print("=" * 60)
```

```python
state = [50, 30]
action = agent.select_action(state)

print("Selected Action Index :", action)
print("Selected Price :", PRICE_LEVELS[action])
```

```python
print("Testing Epsilon-Greedy Action Selection\n")

for i in range(10):
    action = agent.select_action(state)
    print(f"Attempt {i+1}: Action {action} -> Price ₹{PRICE_LEVELS[action]}")
```

```python
agent.remember(state=[50,30],action=2,reward=120,next_state=[49,29],done=False)

agent.remember(state=[49,29],action=1,reward=100,next_state=[48,28],done=False)

agent.remember(state=[48,28],action=4,reward=160,next_state=[47,27],done=False)

print("Replay Buffer Size")
print(len(agent.memory))
```

```python
print("Stored Experiences\n")

for experience in agent.memory.memory:
    print(experience)
```

```python
print("Current Epsilon :", agent.epsilon)

for _ in range(5):
    agent.decay_epsilon()

print("Updated Epsilon :", agent.epsilon)
```

```python
agent.epsilon = 1.0

print("Epsilon Decay Demonstration\n")

print(f"{'Iteration':<12}{'Epsilon'}")

print("-"*25)

for i in range(10):
    print(f"{i+1:<12}{agent.epsilon:.4f}")
    agent.decay_epsilon()
```
