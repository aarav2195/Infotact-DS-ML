```python
state_size = 2

action_size = len(PRICE_LEVELS)

model = DQN(state_size,action_size)

print(model)
```

```python
sample_state = torch.FloatTensor([[50, 30]])

output = model(sample_state)

print("Predicted Q-Values")

print(output)
```

```python
states = torch.FloatTensor([
    [50, 30],
    [40, 20],
    [25, 15],
    [10, 5]
])

predictions = model(states)

print("Predicted Q-Values for Batch States")
print(predictions)
```

```python
print("Batch Predictions\n")

for i in range(len(states)):

    print(f"State {i+1}: {states[i].tolist()}")

    print("Q-Values:")

    print(predictions[i].detach().numpy())

    print("-" * 50)
```
