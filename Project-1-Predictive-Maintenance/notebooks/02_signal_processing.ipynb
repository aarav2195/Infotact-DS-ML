```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"E:\AARAV\Infotact-DS-ML\Project-1-Predictive-Maintenance\data\raw\ai4i2020.csv")

df.head()
```

```python
df["air_temp_mean_10"] = (df["Air temperature [K]"].rolling(window=10).mean())
```

```python
df["process_temp_mean_10"] = (df["Process temperature [K]"].rolling(window=10).mean())
df["rpm_mean_10"] = (df["Rotational speed [rpm]"].rolling(window=10).mean())
df["torque_mean_10"] = (df["Torque [Nm]"].rolling(window=10).mean())
df["tool_wear_mean_10"] = (df["Tool wear [min]"].rolling(window=10).mean())
```

```python
df["air_temp_std_10"] = (df["Air temperature [K]"].rolling(window=10).std())
df["process_temp_std_10"] = (df["Process temperature [K]"].rolling(window=10).std())
df["rpm_std_10"] = (df["Rotational speed [rpm]"].rolling(window=10).std())
df["torque_std_10"] = (df["Torque [Nm]"].rolling(window=10).std())
df["tool_wear_std_10"] = (df["Tool wear [min]"].rolling(window=10).std())
```

```python
df["air_temp_var_10"] = (df["Air temperature [K]"].rolling(window=10).var())
df["process_temp_var_10"] = (df["Process temperature [K]"].rolling(window=10).var())
df["rpm_var_10"] = (df["Rotational speed [rpm]"].rolling(window=10).var())
df["torque_var_10"] = (df["Torque [Nm]"].rolling(window=10).var())
df["tool_wear_var_10"] = (df["Tool wear [min]"].rolling(window=10).var())
```
