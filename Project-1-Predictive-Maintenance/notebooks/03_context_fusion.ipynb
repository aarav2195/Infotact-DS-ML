```python
df["timestamp"] = pd.date_range(start="2026-01-01",periods=len(df),freq="min")

df.head()
```

```python
external_df = pd.DataFrame()

external_df["timestamp"] = df["timestamp"]
```

```python
external_df["ambient_temperature"] = (df["Process temperature [K]"] + np.random.normal(0, 0.5, len(df)))
external_df["load_density"] = (df["Torque [Nm]"] / 100)
```

```python
fused_df = pd.merge(df,external_df,on="timestamp",how="inner")

fused_df.head()
```
