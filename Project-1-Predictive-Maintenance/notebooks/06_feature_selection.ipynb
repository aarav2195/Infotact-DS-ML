```python
target = "Machine failure"

X = df.drop(target,axis=1)
y = df[target]
```

```python
X = X.select_dtypes(include=np.number)
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(n_estimators=200,random_state=42)

model.fit(X,y)
```

```python
importance = pd.DataFrame({"Feature": X.columns, "Importance": model.feature_importances_})

importance = importance.sort_values(by="Importance",ascending=False)

importance
```

```python
top_features = importance.head(15)["Feature"]

top_features
```

```python
final_df = df[list(top_features) + ["Machine failure"]]
```

```python
final_df.to_csv(r"E:\AARAV\Infotact-DS-ML\Project-1-Predictive-Maintenance\data\processed\model_ready_dataset.csv",index=False)
```
