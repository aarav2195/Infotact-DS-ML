```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"E:\AARAV\Infotact-DS-ML\Project-1-Predictive-Maintenance\data\processed\context_fused_dataset.csv")

df.head()
```

```python
df.columns
```

```python
df["temperature_difference"] = (df["Process temperature [K]"] - df["Air temperature [K]"])
```

```python
df["temperature_ratio"] = (df["Process temperature [K]"] / df["Air temperature [K]"])
```

```python
df["load_stress"] = (df["Torque [Nm]"] * df["load_density"])
```

```python
df["rpm_torque_interaction"] = (df["Rotational speed [rpm]"] * df["Torque [Nm]"])
```

```python
df["ambient_temp_mean_10"] = (df["ambient_temperature"].rolling(window=10).mean())
```

```python
df["load_density_mean_10"] = (df["load_density"].rolling(window=10).mean())
```

```python
df = df.dropna()
```
