```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import folium

from folium.plugins import MarkerCluster
from folium.plugins import HeatMap

print("Libraries imported successfully.")
```

```python
df = pd.read_csv("../data/processed/spatial_features.csv")

print(df.shape)

df.head()
```

```python
print("Latitude")

print(df["lat"].describe())

print()

print("Longitude")

print(df["long"].describe())
```

```python
king_map = folium.Map(location=[df["lat"].mean(),df["long"].mean()],zoom_start=10,tiles="CartoDB Positron")

king_map.save("../reports/week-1/visualizations/property_locations.html")

print("Property Location Map Saved Successfully")
```

```python
cluster_map = folium.Map(location=[df["lat"].mean(),df["long"].mean()],zoom_start=10,tiles="CartoDB Positron")

marker_cluster = MarkerCluster().add_to(cluster_map)

for _, row in df.iterrows():
    popup = f"""
    <b>Price:</b> ${row['price']:,.0f}<br>
    Bedrooms: {row['bedrooms']}<br>
    Bathrooms: {row['bathrooms']}<br>
    Living Area: {row['sqft_living']} sqft
    """

    folium.CircleMarker(location=[row["lat"], row["long"]],radius=4,popup=popup,color="blue",fill=True,fill_opacity=0.7).add_to(marker_cluster)

cluster_map.save("../reports/week-1/visualizations/property_locations.html")

print("Marker Cluster Map Saved")
```

```python
heat_data = [
    [
        row["lat"],
        row["long"],
        row["price"]
    ]
    for _, row in df.iterrows()
]

heat_map = folium.Map(location=[df["lat"].mean(),df["long"].mean()],zoom_start=10,tiles="CartoDB Positron")

HeatMap(heat_data,radius=10,blur=15).add_to(heat_map)

heat_map.save("../reports/week-1/visualizations/property_price_heatmap.html")

print("Price Heatmap Saved")
```

```python
price_map = folium.Map(location=[df["lat"].mean(),df["long"].mean()],zoom_start=10,tiles="CartoDB Positron")

sample = df.sample(1000, random_state=42)

for _, row in sample.iterrows():
    if row["price"] >= 700000:
        color = "red"
    elif row["price"] >= 400000:
        color = "orange"
    else:
        color = "green"

    folium.CircleMarker(location=[row["lat"], row["long"]],radius=4,color=color,fill=True,
                        fill_color=color,fill_opacity=0.8,popup=f"${row['price']:,.0f}").add_to(price_map)

price_map.save("../reports/week-1/visualizations/property_price_distribution.html")

print("Price Distribution Map Saved")
```
