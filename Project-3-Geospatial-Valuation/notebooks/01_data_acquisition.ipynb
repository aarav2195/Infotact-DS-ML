```python
import geopandas as gpd
from shapely.geometry import Point

print("GeoPandas Version:", gpd.__version__)
```

```python
df_clean = df.copy()

print(df_clean.shape)
```

```python
missing = pd.DataFrame({
    "Missing Values": df_clean.isnull().sum(),
    "Percentage": round(df_clean.isnull().mean()*100,2)
})

missing
```

```python
print("Duplicate rows:", df_clean.duplicated().sum())

df_clean.drop_duplicates(inplace=True)

print(df_clean.shape)
```

```python
df_clean["date"] = pd.to_datetime(df_clean["date"])

df_clean["Sale Year"] = df_clean["date"].dt.year

df_clean["House Age"] = (
    df_clean["Sale Year"] -
    df_clean["yr_built"]
)
```

```python
Q1 = df_clean["price"].quantile(0.25)

Q3 = df_clean["price"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5*IQR

upper = Q3 + 1.5*IQR

df_clean["price"] = df_clean["price"].clip(
    lower=lower,
    upper=upper
)

print("Price outliers normalized.")
```

```python
plt.figure(figsize=(8,5))

plt.hist(df_clean["price"], bins=40)

plt.title("Normalized House Prices")

plt.xlabel("Price")

plt.ylabel("Count")

plt.show()
```

```python
geometry = [
    Point(xy)
    for xy in zip(
        df_clean["long"],
        df_clean["lat"]
    )
]

gdf = gpd.GeoDataFrame(
    df_clean,
    geometry=geometry,
    crs="EPSG:4326"
)

gdf.head()
```

```python
print(gdf.crs)
```

```python
gdf.plot(figsize=(8,8),markersize=2)

plt.title("Geographical Distribution of Houses")

plt.show()
```

```python
gdf.to_file("../data/processed/clean_houses.geojson",driver="GeoJSON")

gdf.to_csv("../data/processed/clean_house_data.csv",index=False)

print("Processed datasets saved.")
```
