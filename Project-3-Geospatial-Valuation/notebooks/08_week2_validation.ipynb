```python
import os
import pandas as pd
import numpy as np

from joblib import load
```

```python
PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))

DATA_DIR = os.path.join(PROJECT_ROOT,"data","processed","week-2")

MODEL_DIR = os.path.join(PROJECT_ROOT,"models")

PREDICTION_DIR = os.path.join(DATA_DIR,"predictions")

print("DATA DIRECTORY:")
print(DATA_DIR)

print("\nMODEL DIRECTORY:")
print(MODEL_DIR)

print("\nPREDICTION DIRECTORY:")
print(PREDICTION_DIR)
```

```python
required_files = [
    os.path.join(PROJECT_ROOT, "data", "processed", "tabular_features.csv"),
    os.path.join(DATA_DIR, "X_train.csv"),
    os.path.join(DATA_DIR, "X_test.csv"),
    os.path.join(DATA_DIR, "y_train.csv"),
    os.path.join(DATA_DIR, "y_test.csv"),
    os.path.join(DATA_DIR, "model_evaluation.csv"),
    os.path.join(DATA_DIR, "prediction_error_analysis.csv"),
    os.path.join(DATA_DIR, "price_range_error_analysis.csv"),
    os.path.join(PREDICTION_DIR, "model_prediction_comparison.csv"),
    os.path.join(MODEL_DIR, "linear_regression_baseline.pkl"),
    os.path.join(MODEL_DIR, "xgboost_baseline.pkl")
]

validation_results = []

for file_path in required_files:
    validation_results.append({
        "File": os.path.relpath(file_path, PROJECT_ROOT),
        "Exists": os.path.exists(file_path)
    })

file_validation = pd.DataFrame(validation_results)

display(file_validation)
```

```python
tabular_features = pd.read_csv(os.path.join(PROJECT_ROOT,"data","processed","tabular_features.csv"))

X_train = pd.read_csv(os.path.join(DATA_DIR, "X_train.csv"))

X_test = pd.read_csv(os.path.join(DATA_DIR, "X_test.csv"))

y_train = pd.read_csv(os.path.join(DATA_DIR, "y_train.csv"))

y_test = pd.read_csv(os.path.join(DATA_DIR, "y_test.csv"))

evaluation = pd.read_csv(os.path.join(DATA_DIR, "model_evaluation.csv"))

error_analysis = pd.read_csv(os.path.join(DATA_DIR, "prediction_error_analysis.csv"))

price_range_analysis = pd.read_csv(os.path.join(DATA_DIR, "price_range_error_analysis.csv"))

prediction_comparison = pd.read_csv(os.path.join(PREDICTION_DIR,"model_prediction_comparison.csv"))

print("Datasets loaded successfully.")
```

```python
print("Tabular Features:", tabular_features.shape)
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
print("Evaluation:", evaluation.shape)
print("Prediction Comparison:", prediction_comparison.shape)
```

```python
print("Missing values in X_train:")
display(X_train.isnull().sum())

print("\nMissing values in X_test:")
display(X_test.isnull().sum())

print("\nMissing values in y_train:")
display(y_train.isnull().sum())

print("\nMissing values in y_test:")
display(y_test.isnull().sum())
```

```python
numeric_columns = X_train.select_dtypes(include=np.number).columns

train_inf = np.isinf(X_train[numeric_columns]).sum().sum()

test_inf = np.isinf(X_test[numeric_columns]).sum().sum()

print("Infinite values in X_train:", train_inf)
print("Infinite values in X_test:", test_inf)
```

```python
print("Training samples:", len(X_train))
print("Holdout samples:", len(X_test))

print("\nTraining percentage:", round(len(X_train) / (len(X_train) + len(X_test)) * 100, 2), "%")

print("Holdout percentage:", round(len(X_test) / (len(X_train) + len(X_test)) * 100, 2), "%")
```

```python
linear_model_path = os.path.join(MODEL_DIR,"linear_regression_baseline.pkl")
xgb_model_path = os.path.join(MODEL_DIR,"xgboost_baseline.pkl")

linear_model = load(linear_model_path)
xgb_model = load(xgb_model_path)

print("Linear Regression model loaded successfully:")
print(type(linear_model))

print("\nXGBoost model loaded successfully:")
print(type(xgb_model))
```

```python
linear_predictions = linear_model.predict(X_test)
xgb_predictions = xgb_model.predict(X_test)

print("Linear Regression predictions:", len(linear_predictions))
print("XGBoost predictions:", len(xgb_predictions))

print("\nExpected holdout predictions:",len(X_test))
```

```python
prediction_validation = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "XGBoost"
    ],
    "Prediction Count": [
        len(linear_predictions),
        len(xgb_predictions)
    ],
    "Missing Predictions": [
        np.isnan(linear_predictions).sum(),
        np.isnan(xgb_predictions).sum()
    ],
    "Infinite Predictions": [
        np.isinf(linear_predictions).sum(),
        np.isinf(xgb_predictions).sum()
    ]
})

display(prediction_validation)
```

```python
linear_row = evaluation[evaluation["Model"] == "Linear Regression"].iloc[0]

xgb_row = evaluation[evaluation["Model"] == "XGBoost"].iloc[0]

rmse_improvement = ((linear_row["RMSE"] - xgb_row["RMSE"]) / linear_row["RMSE"]) * 100

mape_improvement = ((linear_row["MAPE (%)"] - xgb_row["MAPE (%)"]) / linear_row["MAPE (%)"]) * 100

print(
    f"RMSE improvement with XGBoost: "
    f"{rmse_improvement:.2f}%"
)

print(
    f"MAPE improvement with XGBoost: "
    f"{mape_improvement:.2f}%"
)
```
